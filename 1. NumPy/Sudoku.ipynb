```python
import numpy as np
```

```python
# Valid Sudoku

s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
     
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
s1 = np.sum(s, axis = 1)
```

```python
# Every Row should be a sum of 45

for i in s1:
    if i != 45:
        print(i + "\n")
        print("Row not giving 45")
    else:
        print("Every row is correct")
```

```python
#every 3x3 matrix sum should be 45

for i in range(0, 9, 3):
    for j in range(0, 9, 3):
        s3 = s[i:i+3, j:j+3]
        print(s3.sum())
```

```python
#every Column sum should be 45

s4 = np.sum(s, axis = 0)
for i in s4:
    print(i)
```
