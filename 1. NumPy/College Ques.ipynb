```python
import numpy as np

#Columns: [Age, Math marks, Science marks]

data = np.array([
    [18, 85, 78],    # Student1
    [19, 92, 88],    # Student2
    [17, 76, 95],    # Student3
    [18, 65, 70],    # Student4
    [20, 90, 85]     # Student5
])

size = len(data)
```

```python
#Get the shape of the matrix

data.shape
```

```python
#Find the avg age of students

avgAge = (data[0:, 0].sum())/size
avgAge
```

```python
#Extract math marks of all students

math = data[:,1:2]
math
```

```python
#Find the highest Science marks

science = data[0:, 2:]

highest = 0
for i in science:
    if i > highest:
        highest = i

print("Highest marks = ", highest)
```

```python
#Get details of student who scored more than 90 in math

math = data[0:, 1:2]

for i in range(len(math)):
    if math[i] > 90:
        print(data[i])
```

```python
# Increase math marks of all students by 5

dataCopy1 = data[0:, 0:]

for i in range(size):
    dataCopy1[i][1] += 5

dataCopy1
```

```python
#Find how many students are younger than 19

count = 0

for i in range(size):
    if data[i][0] < 19:
        print(data[i][0])
        count += 1

print("Students younger than 19: ", count)
```

```python
#Calculate avg marks in each subject(Column wise mean)

scienceAvg = science.sum()/size
mathAvg = math.sum()/size

print("Science avg marks: ", scienceAvg)
print("Math avg marks: ", mathAvg)
```

```python
#Get data of student who scored atleast 80 in Both subject

for i in range(size):
    if data[i][1] >= 80 and data[i][2] >= 80:
        print(data[i])
```

```python
#Replace all Science marks les than 75 with 0

dataCopy2 = data[0:, 0:]

for i in range(size):
    if dataCopy2[i][2] < 75:
        dataCopy2[i][2] = 0
        
dataCopy2
```
