```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1, 10, 20)
x
```

```python
y = x**2
y
```

```python
plt.plot(x, y)
```

```python
z = y**8
```

```python
plt.plot(z, x)
```

```python
plt.title("No IDEA!!!")
plt.xlabel("z")
plt.ylabel("y")
plt.plot(z, y)
```

```python
plt.subplot(2, 3, 5)
plt.plot(x, y)
plt.subplot(2, 3, 1)
plt.plot(z, y)
plt.subplot(2, 3, 3)
plt.plot(y, x)
plt.subplot(2, 3, 6)
plt.plot(x, x)
# plt.subplot(2, 3, 7)
```

```python
fig = plt.figure(figsize=(10,8), dpi=100)
axis1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.2, .4, 0.3, .3])
axis2.plot(y,x)
```

```python
plt.scatter(x, y)
```

```python
fig = plt.figure()
axes1 = fig.add_axes([.3, .2, .7, .7])
axes1.set_xlabel("x-axis")
axes1.set_ylabel("y-axis")
axes1.plot(x, y, color = "red")
plt.savefig("firstFig.png")
```
