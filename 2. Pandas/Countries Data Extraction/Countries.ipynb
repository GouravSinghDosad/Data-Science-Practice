```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.read_csv('Countries.csv')
```

```python
df1
```

```python
# Which Country has the highest population

hPop = df1["population"].idxmax()
locHPop = df1.loc[hPop]
print("Country with the highest population: " , locHPop["country"])
```

```python
# What is the capital of the country with the highest population

locHPopCapital = locHPop["capital_city"]
print("Capital of the country with the highest population: ", locHPopCapital)
```

```python
# Which Country has the least population

lPop = df1["population"].idxmin()
locLPop = df1.loc[lPop]
print("Country with the least population: ", locLPop["country"])
```

```python
# What is the capital of the country with the least population

locLPopCapital = locLPop["capital_city"]
print("capital of the country with the least population: ", locLPopCapital)
```

```python
# Give me top 5 countries with the highest democratic score

df2 = df1.sort_values(by= "democracy_score", ascending= False)
df2[["democracy_score", "country"]].head(5)
```

```python
# How many total regions are there(unique)

df1UniqueRegion = pd.DataFrame(df1["region"].unique())
df1UniqueRegion
```

```python
# How many countries lie in eastern europe region 

regions = pd.DataFrame(df1["region"].value_counts())
regions
regions.loc["Eastern Europe"]
```

```python
df1[df1["region"] == "Eastern Europe"]["country"]
```

```python
#Who is the political Leader of the 2nd highest populated Country

populationInDESC = df1.sort_values(by= "population", ascending= False)

print("political Leader of the 2nd highest populated Country is - ", populationInDESC.iloc[1]["political_leader"])
```

```python
# How many countries are there whoes political leaders are unknown

df1[df1["political_leader"].isnull()]["country"]
```

```python
print("Number of Unkown Political Leader countries: ", df1[df1["political_leader"].isnull()]["country"].count())
```

```python
# How many country have Republic in their full name

df1[df1["country_long"].str.contains("Republic", na= False)]
```

```python
#Which country in african region has highest population

africanCountries = df1[df1["continent"] == "Africa"]
africanCountries.sort_values(by = "population",ascending= False, inplace= True)

print("Highly populated country in africa: ", africanCountries["country"].iloc[0])
```
