```python
# Group BY

df7 = pd.DataFrame({
    'category': ['A', 'B', 'A', 'B', 'C'],
    "sales": [12000, 8900, 12000, 230, 12000],
    "hike": [1000, 3000, 1000, 3000, 2000]
})
```

```python
df7
```

```python
df8 = df7.groupby("category")
```

```python
for a, b in df8:
    print(a)
    print(b)
```

```python
df7.groupby("category")['sales'].sum()
df9 = df7.groupby('hike')
```

```python
for i, v in df9:
    print(i)
    print(v)
```

```python
df9['sales'].sum()
```

```python
df10 = df7.groupby(['category', 'sales'])['hike'].sum()
df10
```
