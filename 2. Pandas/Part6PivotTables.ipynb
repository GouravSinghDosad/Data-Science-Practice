```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': [
        'East', 'West', 'North', 'South',
        'East', 'West', 'North', 'South',
        'East', 'West', 'North', 'South',
        'East', 'West', 'North', 'South',
        'East', 'West', 'North', 'South'
    ],
    'Sales': np.random.randint(1000, 10000, 20),
    'Units': np.random.randint(10, 200, 20),
    'Rep': [
        'John', 'Mary', 'Bob', 'Alice',
        'John', 'Mary', 'Bob', 'Alice',
        'John', 'Mary', 'Bob', 'Alice',
        'John', 'Mary', 'Bob', 'Alice',
        'John', 'Mary', 'Bob', 'Alice'
    ]
}

df = pd.DataFrame(data)
```

```python
df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = "Q" + df['Date'].dt.quarter.astype(str)
```

```python
df
```

```python
pd.pivot_table(df, values= 'Units', index= 'Rep', columns= 'Product')
```

```python
pd.pivot_table(df, values= 'Units', index= 'Rep', columns= 'Product', aggfunc='median')
```

```python
pd.pivot_table(df, values= ['Sales', 'Units'], index= 'Rep', columns= 'Product', aggfunc='median')
```
