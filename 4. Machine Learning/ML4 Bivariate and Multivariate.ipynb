```python
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
import plotly as pl
```

```python
titanic = pd.read_csv('titanic.csv')
```

```python
flights = sb.load_dataset('flights')
```

```python
tips = sb.load_dataset('tips')
```

```python
iris = sb.load_dataset('iris')
```

### 1. ScatterPlot (Numerical - Numerical)

```python
sb.scatterplot(x= 'sepal_length', y= 'sepal_width', data= iris, hue='species')
```

```python
import plotly.express as px

fig = px.scatter(
    iris, 
    x="sepal_length", 
    y="sepal_width", 
    color="species"   # equivalent to hue in Seaborn
)
fig.show()
```

```python
sb.scatterplot(x = iris['sepal_length'], y = iris['sepal_width'], hue=iris['species'], style= iris['petal_width'])
```

```python
sb.scatterplot(x= 'total_bill', y= 'tip', hue= 'day', style= 'sex', data=tips)
```

### 2. BarPlot (Numerical - Categorical)

```python
sb.barplot(y= 'Age', x= 'Pclass', data= titanic, hue='Embarked')
```

```python
titanic.groupby(['Pclass', 'Embarked'])['Age'].mean()
```

## 3. Box Plot (Numerical - Categorical)

```python
sb.boxplot(x= 'month', y= 'passengers', data= flights, hue='month')
```

```python
sb.boxenplot(x= 'month', y= 'passengers', data= flights, hue='month')
```

```python
sb.boxplot(x= 'sex', y= 'tip', data= tips, hue='day')
```

```python
import plotly.express as px

fig = px.box(
    tips, 
    x="sex", 
    y="tip", 
    color="day"   
)
fig.show()
```

## 4. DistPlot (Numerical - Categorical)

```python
titanic.head()
```

```python
sb.distplot(titanic['Survived'])
```

```python
sb.distplot(titanic[titanic['Survived']==0]['Age'], hist= False, label= 'died')
sb.distplot(titanic[titanic['Survived']==1]['Age'], hist= False, label= 'survived')
plt.legend()
```

## 5. Heatmap (Categorical - Categorical)

```python
titanic.head()
```

```python
corr1 = titanic.corr(numeric_only= True)
```

```python
sb.heatmap(corr1, annot= True)
```

```python
corr2 = pd.crosstab(titanic['Pclass'], titanic['Survived'])
corr2
```

```python
sb.heatmap(corr2, annot= True)
```

```python
corr3 = titanic.groupby('Pclass')['Survived'].mean() * 100
corr3.plot(kind= 'bar')
```

## 6. ClusterMap (Categorical - Categorical)

```python
sb.clustermap(pd.crosstab(tips['tip'], tips['time']))
```

## 7. PairPlot

```python
sb.pairplot(tips, hue='day')
```

```python
sb.pairplot(iris, hue='species')
```
