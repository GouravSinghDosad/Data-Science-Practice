```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
df = pd.read_csv('titanic.csv')
df.head()
```

## 1.Categorical Data

### a. CountPlot

```python
sb.countplot(x = 'Survived', data=df, hue='Sex')
```

```python
sb.countplot(y = 'Survived', data=df, hue='Survived')
```

```python
df['Survived'].value_counts()
```

### c.Pie Chart

```python
df['Survived'].value_counts().plot(kind= 'pie', autopct= '%.2f')
```

```python
df['Embarked'].value_counts().plot(kind= 'pie', autopct= '%.2f')
```

## 2.Numerical Data

### a.Histogram

```python
plt.hist(df['Age'], bins= 8)
```

### b. Distplot

```python
sb.distplot(df['Age'])
```

### c. Boxplot

```python
sb.boxplot(x = 'Sex', y='Age', data=df)
```
