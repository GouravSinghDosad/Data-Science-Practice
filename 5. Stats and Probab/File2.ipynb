```python
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
```

# Feature Selection (1D plot)

```python
Iris = pd.read_csv('Iris.csv')
```

```python
Iris.sample(10)
```

```python
sb.kdeplot(x= 'PetalLengthCm', hue="Species", data= Iris)
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12,10))

ax1, ax2, ax3, ax4 = axes.flatten()

sb.kdeplot(data=Iris, x="SepalLengthCm", hue="Species", ax=ax1)
sb.kdeplot(data=Iris, x="SepalWidthCm", hue="Species", ax=ax2)
sb.kdeplot(data=Iris, x="PetalLengthCm", hue="Species", ax=ax3)
sb.kdeplot(data=Iris, x="PetalWidthCm", hue="Species", ax=ax4)

plt.show()
```

```python
Titanic = pd.read_csv('titanic.csv')
Titanic.sample(5)
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize= (12, 10))

ax1, ax2, ax3, ax4= axes.flatten()

sb.kdeplot(data= Titanic[Titanic['Survived'] == 1], x='Age' , hue= Titanic['Embarked'], ax= ax1)
sb.kdeplot(data= Titanic[Titanic["Survived"] == 0], x= "Age", hue= Titanic["Embarked"], ax= ax2)
sb.kdeplot(data= Titanic[Titanic["Survived"] == 0], x= "Age", hue= Titanic["Pclass"], ax= ax3)
sb.kdeplot(data= Titanic[Titanic["Survived"] == 0], x= "Age", hue= Titanic["Pclass"], ax= ax4)
```

```python
sb.kdeplot(data= Titanic, x= "Age", hue= "Survived")
```

```python
sb.kdeplot(data= Iris, x= "PetalWidthCm", hue= "Species")
sb.ecdfplot(data= Iris, x= "PetalWidthCm", hue= "Species")
```
